```python
ntot = 100
P = 15
N = ntot - P

FP_list = list(range(N + 1))
FN_list = list(range(P + 1))

def acc(FP, FN, TP, TN):    
    return (TP + TN) / (TP + TN + FP + FN)

def jaccard(FP, FN, TP):    
    return TP / (TP + FP + FN)

acc_list = []
jaccard_list = []
for FP in FP_list:
    TN = N- FP
    for FN in FN_list:
        TP = P - FN
        acc_list.append(acc(FP, FN, TP, TN))
        jaccard_list.append(jaccard(FP, FN, TP))
        
    
```

```python
from matplotlib import pyplot as plt

plt.scatter(acc_list, jaccard_list)
```
